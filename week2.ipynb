```python
#Pre-Lab_Q1
#Write a Python program to perform basic arithmetic operations and statistical calculations.
import statistics

numbers=[10,20,30,40,50]

print("Numbers:",numbers)

print("Addition:",sum(numbers))
print("Subtraction:",numbers[1]-numbers[0])
print("Multiplication:",numbers[0]*numbers[1])
print("Division:",numbers[4]/numbers[0])

print("Mean:",statistics.mean(numbers))
print("Median:",statistics.median(numbers))
print("Maximum:",max(numbers))
print("Minimum:",min(numbers))
```

```python
#Pre_Lab_Q3
#Write a Python program to load and analyze a CSV file using Pandas.
import pandas as pd

df=pd.read_csv("students.csv")

print(df)

print("\nShape")
print(df.shape)

print("\nInformation")
print(df.info())

print("\nSummary Statistics")
print(df.describe())

print("\nAverage Marks")
print(df["Marks"].mean())
```

```python
# In-Lab
#Write a Python program for Python Fundamentals for Data Analytics & Introduction to NumPy and Pandas.

import numpy as np
import pandas as pd

student={
"ID":[101,102,103,104,105],
"Name":["Rahul","Sneha","Ravi","Anitha","Priya"],
"Marks":[85,92,78,88,95]
}

df=pd.DataFrame(student)

print("Student Dataset")
print(df)

marks=np.array(df["Marks"])

print("\nNumPy Array")
print(marks)

print("\nStatistics")
print("Total Marks:",np.sum(marks))
print("Average Marks:",np.mean(marks))
print("Maximum Marks:",np.max(marks))
print("Minimum Marks:",np.min(marks))
print("Standard Deviation:",round(np.std(marks),2))

print("\nStudents Scoring Above 80")
print(df[df["Marks"]>80])

print("\nSummary Statistics")
print(df.describe())
```
