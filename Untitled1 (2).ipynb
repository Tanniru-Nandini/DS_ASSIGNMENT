```python
import dash
from dash import html, dcc

app = dash.Dash(__name__)

app.layout = html.Div([
    html.H1("Basic Dashboard"),

    html.Label("Select an option:"),

    dcc.Dropdown(
        options=[
            {"label": "Option 1", "value": "1"},
            {"label": "Option 2", "value": "2"},
            {"label": "Option 3", "value": "3"}
        ],
        value="1"
    ),

    html.Br(),

    html.H3("Welcome to my Dashboard"),

    dcc.Graph(
        figure={
            "data": [
                {
                    "x": [1, 2, 3, 4, 5],
                    "y": [2, 4, 6, 8, 10],
                    "type": "line",
                    "name": "Line"
                }
            ],
            "layout": {
                "title": "Basic Line Graph",
                "xaxis": {"title": "X Values"},
                "yaxis": {"title": "Y Values"}
            }
        }
    )
])

if __name__ == "__main__":
    app.run(debug=True)
```
