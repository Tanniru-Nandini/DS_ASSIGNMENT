```python
#sum in numpy
import numpy as np

arr = np.array([1, 2, 3, 4])

total = np.sum(arr)
print(total)
```

```python
#row sum
arr = np.array([[1, 2],
                [3, 4]])
row_sum = np.sum(arr, axis=1)
print(row_sum)
```

```python
#col sum
col_sum = np.sum(arr, axis=0)
print(col_sum)
```

```python
#looping for 1D
a = np.array([10, 20, 30, 40])

for x in a:
    print(x)
```

```python
#looping for 2D
import numpy as np

a = np.array([[1, 2, 3],
              [4, 5, 6]])

for row in a:
    print(row)
```
