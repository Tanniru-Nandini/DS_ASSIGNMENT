```python
import matplotlib.pyplot as plt

y1 = []
y2 = []

x=range(-100,100,10)

for i in x:
    
    y1.append(i**2)
    y2.append(-i**2)

plt.plot(x, y1)
plt.plot(x, y2)
plt.title("simple graph")
plt.xlabel("x-label")
plt.ylabel("y-label")
plt.show()
```

```python
import matplotlib.pyplot as plt

y1 = []
y2 = []

x=range(-100,100,10)

for i in x:
    
    y1.append(i**2)
    y2.append(-i**2)

plt.plot(x, y1,color='blue')
plt.plot(x, y2,color='green')
plt.title("simple graph")
plt.xlabel("x-label")
plt.ylabel("y-label")
plt.axhline(color='pink')
plt.axvline(color='red')

plt.show()
```
