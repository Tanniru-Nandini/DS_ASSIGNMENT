```python
import numpy as np
from scipy.spatial import distance
pointA=np.array([2,4,6])
pointB=np.array([5,1,9])
euclidean_dist=distance.euclidean(pointA,pointB)
print("Euclidean Distance:",euclidean_dist)
simarility_euclidean=1/(1+euclidean_dist)
print("Euclidean Similarity:",simarility_euclidean)
```

```python
pointA=np.array([9,7,5])
pointB=np.array([3,2,8])
manhattan_dist=distance.cityblock(pointA,pointB)
print("Manhattan  Distance:",manhattan_dist)
simarility_manhattan=1/(1+manhattan_dist)
print("manhattan Similarity:",simarility_manhattan)
```

```python
pointA=np.array([9,7,5])
pointB=np.array([3,2,8])
minkowski_dist_p3=distance.minkowski(pointA,pointB,p=3)
print("Minkowski  Distance(p=3):",minkowski_dist_p3)
simarility_minkowski=1/(1+minkowski_dist_p3)
print("Minkowski Similarity:",simarility_minkowski)
```

```python
def hamming_distance(s1, s2):
    if len(s1) != len(s2):
        return ValueError("string must be of equal length")

    return sum(ch1 != ch2 for ch1, ch2 in zip(s1, s2))


s1 = "karolin"
s2 = "kathrin"

dist = hamming_distance(s1, s2)
print(f"Hamming distance between {s1} and {s2}: {dist}")
```

```python
def jaccard_index(str1, str2):
    set1,set2 = set(str1.split()),set(str2.split())
 

    intersection = set1.intersection(set2)
    union = set1.union(set2)

    return len(intersection) / len(union)


s1 = "data science is fun"
s2 = "science makes data useful"

jaccard = jaccard_index(s1, s2)

print(f"Jaccard Index: {jaccard}")
```

```python
def lcs_length(X, Y):
    m, n = len(X), len(Y)
    dp = [[0] * (n + 1) for _ in range(m + 1)]

    for i in range(m):
        for j in range(n):
            if X[i] == Y[j]:
                dp[i + 1][j + 1] = dp[i][j] + 1
            else:
                dp[i + 1][j + 1] = max(dp[i][j + 1], dp[i + 1][j])

    return dp[m][n]


seq1 = "ABCDEF"
seq2 = "AEBDF"

l = lcs_length(seq1, seq2)

print(f"Longest Common subsequence length between '{seq1}' and '{seq2}' : {l}")
```
